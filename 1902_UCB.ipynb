```python
M = int(input("enter number of machines"))
machines = []
for i in range(M):
    name = input("enter machine name")
    machines.append(name)

print(machines)
```

```python
import numpy as np
import math as mt

N_play = 5

N = np.zeros((M, N_play))
R = np.zeros((M, N_play))
Q = np.zeros((M, N_play))
UCB = [0, 0, 0]

Q_old = 0
for machine in machines:
    print("Play with machine", machine)
    r = int(input("enter reward"))

    id = machines.index(machine)

    N[id, 0] = 1
    R[id, 0] = r

    Q[id, 0] = Q_old + (R[id, 0] - Q_old) / N[id, 0]

print(N)
print(R)
print(Q)
```

```python
# UCB
import math as mt
play = 1
c = 2
t = 3
while(play < N_play-1):
    t = t + play
    Q[:, play] = Q[:, play-1]

    for id in range(M):

        UCB[id] = Q[id, play-1] + c * mt.sqrt( mt.log(t) / np.sum(N[id, :]))
    print(UCB)
    id_max = np.argmax(UCB)

    print("Play with machine", machines[id_max])

    r = int(input("enter reward"))

    r = int(r)

    N[id_max, play] = 1

    R[id_max, play] = r

    Q[id_max, play] = Q[id_max, play-1] + (r - Q[id_max, play-1]) / np.sum(N[id_max, :])

    play = play + 1


print(N)
print(R)
print(Q)
```
