```python
import gymnasium as gym

env = gym.make("CartPole-v1")

state, info = env.reset()

total_reward = 0
steps = 0

while True:
    action = env.action_space.sample()

    next_state, reward, terminated, truncated, info = env.step(action)

    print("State:", state)
    print("Action:", action)
    print("Reward:", reward)
    print("Next State:", next_state)
    print()

    total_reward += reward
    steps += 1

    state = next_state

    if terminated or truncated:
        break

average_reward = total_reward / steps

print("Total Reward:", total_reward)
print("Number of Steps:", steps)
print("Average Reward:", average_reward)

env.close()
```

FrozenLake

```python
import gymnasium as gym


def random_agent(env_name):


    if env_name == "FrozenLake-v1":
        env = gym.make(env_name, is_slippery=False)
    else:
        env = gym.make(env_name)


    state, info = env.reset()

    total_reward = 0
    steps = 0

    print("\nEnvironment:", env_name)
    print("Initial State:", state)
    print("Action Space:", env.action_space)
    print("--------------------------------")

    while True:


        action = env.action_space.sample()

        next_state, reward, terminated, truncated, info = env.step(action)

        print("State:", state)
        print("Action:", action)
        print("Reward:", reward)
        print("Next State:", next_state)
        print()

        total_reward += reward
        steps += 1
        state = next_state


        if terminated or truncated:
            break


    average_reward = total_reward / steps

    print("--------------------------------")
    print("Total Reward:", total_reward)
    print("Number of Steps:", steps)
    print("Average Reward:", average_reward)

    env.close()


random_agent("CartPole-v1")
random_agent("MountainCar-v0")
random_agent("FrozenLake-v1")
```

```python
import gymnasium as gym
env=gym.make("CartPole-v1")
#reset environment
current_state,info=env.reset()
print("Cart Position:",current_state[0])
print("Cart Velocity:",current_state[1])
print("Pole Angle:",current_state[2])
print("Pole Angular Velocity:",current_state[3])
#random action
action=env.action_space.sample()
if(action==0):
    print("Action:Left")
else:
    print("Action:Right")
next_state,reward,terminated,truncated,info=env.step(action)
print("Next Cart Position:",next_state[0])
print("Next Cart Velocity:",next_state[1])
print("Next Pole Angle:",next_state[2])
print("Next Pole Angular Velocity:",next_state[3])
print("Reward:",reward)
print("Terminated:",terminated)
print("Truncated:",truncated)
print("Info:",info)
```
