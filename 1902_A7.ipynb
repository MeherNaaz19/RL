Undiscounted and Discounted Returns

```python

n = int(input("Enter the number of time steps: "))

rewards = []

for i in range(n):
    reward = float(input(f"Enter reward at time step {i + 1}: "))
    rewards.append(reward)

undiscounted_return = sum(rewards)

gamma = float(input("Enter the discount factor (gamma): "))

discounted_return = 0

for t in range(n):
    discounted_return += (gamma ** t) * rewards[t]

print("\nEntered reward values:", rewards)
print("Undiscounted return:", undiscounted_return)
print("Discounted return:", discounted_return)
```

2 × 3 Grid World MDP

```python

class GridWorld:
    def __init__(self):

        self.states = [
            (0, 0), (0, 1), (0, 2),
            (1, 0), (1, 1), (1, 2)
        ]

        self.actions = ["up", "down", "left", "right"]

        self.goal = (1, 2)
        self.obstacle = (1, 1)

        self.gamma = 0.9


    def reward(self, state):
        if state == self.goal:
            return 10
        elif state == self.obstacle:
            return -10
        else:
            return -1


    def transition(self, state, action):
        row, col = state

        if action == "up":
            next_state = (row - 1, col)

        elif action == "down":
            next_state = (row + 1, col)

        elif action == "left":
            next_state = (row, col - 1)

        elif action == "right":
            next_state = (row, col + 1)

        else:
            return state


        if next_state not in self.states:
            return state

        if next_state == self.obstacle:
            return state

        return next_state



if __name__ == "__main__":

    env = GridWorld()

    print("State Space:")
    print(env.states)

    print("\nAction Space:")
    print(env.actions)

    print("\nDiscount Factor:")
    print(env.gamma)

    print("\nRewards:")
    for state in env.states:
        print(state, ":", env.reward(state))


    current_state = (0, 0)
    action = "right"

    next_state = env.transition(current_state, action)

    print("\nCurrent State:", current_state)
    print("Action:", action)
    print("Next State:", next_state)
```
